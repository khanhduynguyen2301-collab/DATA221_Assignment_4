```python
import pandas as pd

from sklearn.datasets import load_breast_cancer


cancer_data = load_breast_cancer()

X = cancer_data.data
y = cancer_data.target

print("Shape of X:", X.shape)
print("Shape of y:", y.shape)
print()

# Count samples in each class
class_counts = pd.Series(y).value_counts().sort_index()
print("Class counts:")
for i, count in class_counts.items():
    print(f"{cancer_data.target_names[i]}: {count}")

# In Python comments:
# The dataset is somewhat imbalanced, but not severely imbalanced.
# There are more benign cases than malignant cases.
# Class balance matters because a model can look accurate overall while
# still performing poorly on the minority class, which is especially risky
# in medical classification tasks.
```
